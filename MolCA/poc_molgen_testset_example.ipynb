# SMol-Instruct Forward synthesis dataset

* Input-Output: : `Instruction + Reactant -> Product`
* Total **1K instances**
* Fields in Each instance
  * `graph`: graph of input molecule
  * `input_mol_selfies`: input molecule as SELFIES
  * `input_mol_smiles`: input molecule as SMILES
  * `label_selfies`: label as SELFIES
  * `label_smiles`: label as SMILES
  * `task_subtask_pair`: task specification
  * `instruction`: natural language description to guide LLM which task to solve
  * `condition`: To be filled by property prediction model inference
  
`condition` field를 물성 예측 모델 예측값으로 넣어서 주시면 될 것 같습니다.

# SMol-Instruct Molecule generation dataset

* Input-Output : `Text description of target molecule -> target molecule`
* Total **1K instances**
* Fields in Each instance
  * `graph`: graph of input molecule
  * `input_mol_selfies`: "<None>", as this task has no input molecule
  * `input_mol_smiles`: "<None>", as this task has no input molecule
  * `label_selfies`: label as SELFIES
  * `label_smiles`: label as SMILES
  * `task_subtask_pair`: task specification
  * `instruction`: natural language description to guide LLM which task to solve. In text2mol task, instruction include target molecule text description (text conditioning)
  * `condition`: To be filled by property prediction model inference
  

`condition` field를 물성 예측 모델 예측값으로 넣어서 주시면 될 것 같습니다.
